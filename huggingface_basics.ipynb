```python
from transformers import pipeline
```

```python
classifier = pipeline("sentiment-analysis")

classifier("I love learning AI!")
```

```python
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

generator(
    "Artificial Intelligence is",
    max_new_tokens=50
)
```

```python
generator(
    "Robotics will",
    max_new_tokens=50
)
```
