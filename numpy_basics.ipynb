```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print(arr)
```

```python
print("Shape:", arr.shape)
print("Dimensions:", arr.ndim)
print("Size:", arr.size)
print("Data type:", arr.dtype)
```

```python
print(arr[0])      # First element
print(arr[2])      # Third element
print(arr[1:4])    # Elements from index 1 to 3
print(arr[:3])     # First three elements
print(arr[-2:])    # Last two elements
```

```python
print(arr + 10)
print(arr * 2)
print(arr ** 2)
```

```python
print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
print("Sum:", np.sum(arr))
```

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(matrix)
```

```python
numbers = np.arange(12)

print(numbers)

print(numbers.reshape(3,4))
```

```python
random_numbers = np.random.rand(2,4)

print(random_numbers)
```

```python
arr = np.array([2,5,8,1,9])

print(arr[arr > 5])
```

```python
import pandas as pd
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [20, 21, 22, 23],
    "Marks": [85, 90, 78, 88]
}

df = pd.DataFrame(data)

df
```

```python
print(df)

df.head()

df.tail()
```

```python
df[["Name", "Marks"]]
```

```python
df[df["Marks"] > 80]
df[df["Age"] > 21]
```

```python
print(df["Marks"].mean())
print(df["Marks"].max())
print(df["Marks"].min())
```
