```python
# Detection Metrics

1. IoU
Measures how much the predicted bounding box overlaps the actual bounding box.
Higher IoU means the predicted box is closer to the ground truth.
Values range from 0 to 1.
2. Precision
Out of all the objects the model detected, how many were actually correct?
High precision means fewer false positives.
3. Recall
Out of all the actual objects, how many did the model detect?
High recall means fewer missed detections.
4. mAP@0.5
Mean Average Precision calculated using an IoU threshold of 0.5.
A prediction is considered correct if IoU ≥ 0.5.
Higher values are better.
5. mAP@0.5:0.95
Average mAP calculated from IoU thresholds 0.50 to 0.95.
This is much stricter than mAP@0.5.
It is the main metric used in YOLO research papers.
```
